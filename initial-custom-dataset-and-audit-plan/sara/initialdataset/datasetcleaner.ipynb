```python
import pandas as pd
from IPython.display import FileLink
```

```python
#Load the sheet
```

```python
excel_file = pd.ExcelFile('initial_tiktok_dataset_hashtags.xlsx')
print(excel_file.sheet_names)
```

```python
df = excel_file.parse('Data')
```

```python
df.head()
```

```python
renamed_df = df.rename(columns={
    'authorMeta/fans': 'Follower Count', 
    'authorMeta/heart': 'Total Likes',
    'authorMeta/name': 'Creator Username',
    'shareCount': 'Post Share Count',
    'text': 'Post Caption',
    'hashtags/1/name': 'Hashtag',
    'hashtags/0/title': 'Post Title',
    'authorMeta/signature': 'Creator Bio'
})
```

```python
df_subset = renamed_df[['Hashtag', 'Post Title', 'Post Caption', 'Creator Username','Creator Bio','Total Likes','Follower Count']]
```

```python
df_subset.head()
```

```python
df_subset.to_csv('tiktok_data_cleaned.csv', index=False)
```
